# ctxlab run analysis

Reads `runs/<id>/records.jsonl` (the source of truth) and pivots mean EM / F1 by arrangement and model.

```python
from pathlib import Path
from ctxlab.report import aggregate, report
from ctxlab.tracking import load_records

RUN = Path("runs/smoke")  # change to runs/position_sweep etc.
records = load_records(RUN)
print(f"{len(records)} records")
report(RUN)
```

```python
agg = aggregate(records)
rows = [
    {"arrangement": a, "model": m, **metrics}
    for (a, m), metrics in sorted(agg.items())
]
rows
```
